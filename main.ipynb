```python
# create bronze, silver and gold schemas (catalogs) in DuckDB
con = duckdb.connect("stocks.duckdb")
for catalog in ["bronze", "silver", "gold"]:
    con.execute(f"CREATE SCHEMA IF NOT EXISTS {catalog}")

# verify
print(con.execute("SHOW SCHEMAS").fetchall())
```
